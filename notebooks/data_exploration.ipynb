# Twitter Customer Support: Dataset Exploration & Taxonomy Discovery
This notebook explores the historical customer support interactions for **@AmazonHelp**, audits data cleanliness, checks conversational turn dynamics, and validates the leakage prevention split between the retrieval corpus and evaluation sets.

```python
import os
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
print("Libraries loaded successfully.")
```

## 1. Loading Raw Support Interactions

```python
data_path = "../data/raw/amazon_support_sample.csv"
if not os.path.exists(data_path):
    data_path = "data/raw/amazon_support_sample.csv"
df = pd.read_csv(data_path)
print(f"Loaded {len(df):,} support interaction turns.")
display(df.head(3))
```

## 2. Intent Distribution in Customer Support Inquiries

```python
intent_counts = df["intent"].value_counts()
print("Intent Breakdown:")
display(intent_counts)

plt.figure(figsize=(10, 5))
intent_counts.plot(kind="barh", color="#2563eb")
plt.title("Distribution of Support Inquiries by Discovered Intent", fontsize=13, fontweight="bold")
plt.xlabel("Interaction Count")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## 3. Message Length & Character Distribution

```python
df["cust_len"] = df["customer_message"].apply(len)
df["resp_len"] = df["historical_response"].apply(len)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df["cust_len"], bins=25, ax=axes[0], color="#0284c7", kde=True)
axes[0].set_title("Customer Tweet Length (characters)")
sns.histplot(df["resp_len"], bins=25, ax=axes[1], color="#10b981", kde=True)
axes[1].set_title("Brand Response Length (characters)")
plt.tight_layout()
plt.show()

print(f"Mean Customer Inquiry Length: {df['cust_len'].mean():.1f} chars")
print(f"Mean Historical Response Length: {df['resp_len'].mean():.1f} chars")
```
